```python
import pandas as pd
```

```python
df = pd.read_csv("finance_loan_portfolio_python_dataset.csv")
```

```python
df.isnull().sum()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

```python
df["Annual_Income_INR"] = df["Annual_Income_INR"].fillna(df["Annual_Income_INR"].mean())
```

```python
df["Gender"] = df["Gender"].fillna("Unknown Gender")
```

```python
df["City"] = df["City"].fillna("Unknown City")
```

```python
df["Employment_Type"] = df["Employment_Type"].fillna("Unknown Employment_Type")
```

```python
df["Credit_Score"] = df["Credit_Score"].fillna(df["Credit_Score"].mean())
```

```python
df["Interest_Rate_Pct"] = df["Interest_Rate_Pct"].fillna(df["Interest_Rate_Pct"].mean())
```

```python
df["Outstanding_Balance_INR"] = df["Outstanding_Balance_INR"].fillna(df["Outstanding_Balance_INR"].mean())
```

```python
df["Debt_to_Income_Ratio"] = df["Debt_to_Income_Ratio"].fillna(df["Debt_to_Income_Ratio"].mean())
```

```python
df["Savings_Balance_INR"] = df["Savings_Balance_INR"].fillna(df["Savings_Balance_INR"].mean())
```

```python
df["Investment_Value_INR"] = df["Investment_Value_INR"].fillna(df["Investment_Value_INR"].mean())
```

```python
df["Last_Payment_Date"] = df["Last_Payment_Date"].fillna("Unknown Last_Payment_Date")
```

```python
df["Payment_Mode"] = df["Payment_Mode"].fillna("Unknown Payment_Mode")
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```
